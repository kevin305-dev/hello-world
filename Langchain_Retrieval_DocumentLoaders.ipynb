## Document Loaders

> 다양한 형식의 문서를 불러오고,
> 이를 Langchain에 결합하기 쉬운 텍스트 형태로 변환하는 기능을 합니다.

이를 통해 사용자는
-  txt, pdf, word, ppt, xlsx, csv 등

## 1) URL Document Loader

> 웹에 기록된 글도 텍스트 형식으로 가져와 LLM에 활용할 수 있다.
-  대표적인 URLLoader는 WebBaseLoader, UnstructuredURLLoader 가 있다.

>  WebBaseLoader는 웹페이지의 모든 정보를 가져와 원하지않는 정보를 가져와 정리가 필요함

```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader("https://n.news.naver.com/mnews/article/092/0002307222?sid=105")

data = loader.load()  # list값으로 가져온다
print(data[0].page_content) # list.page_content를 못 불러옴
```

>  UnstructuredURLLoader

```python
from langchain_community.document_loaders import UnstructuredURLLoader

url = [
    "https://n.news.naver.com/mnews/article/002/0002307222?sid=105",
    "https://n.news.naver.com/mnews/article/052/0001944792?sid=105",
]

loader = UnstructuredURLLoader(urls=url)

data = loader.load()
data
```

## CSV Document Loader
-  열별로 값들이 어떻게 입력되어있는지 확인
-  CSVLoader를 통해 로딩한 값들을 SQL과 연동해서 확인 가능
>  LLM을 통해서 SQL을 실행하고 Data 값들을 추출할 수 있다

```python
from langchain_community.document_loaders import CSVLoader

loader = CSVLoader(file_path='/content/drive/MyDrive/폐기물차량리스트1.csv', csv_args={
    'delimiter' : ',',
    'quotechar' : '"',
    'fieldnames' : ['장비MAC주소', '차량번호', '차량구역', '차량제원', '차량종류', '수정일']
}, encoding='cp949')

data = loader.load()
```
